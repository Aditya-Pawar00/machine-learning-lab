# Automobile Price Prediction using Multiple Linear Regression


```python
import pandas as pd
import pickle

with open("model.pkl", "rb") as f:
    model = pickle.load(f)

sample = pd.DataFrame([{"engine-size": 130, "horsepower": 110, "curb-weight": 2500, "highway-mpg": 30, "body-style": "sedan"}])
print("Predicted Price:", model.predict(sample)[0])
```
